```python
# Importing python library 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt  

print("*"*80)
# Loading dataset
df = pd.read_csv("../Datasets/amazon_delivery.csv")
print("Columns Present in the dataset are: ")
print(df.columns.tolist())
```

```python
print("first five rows:")
df.head()

```

```python
df.isnull().sum() #checking the missing values 
```

```python
# Agent Rating is missing in 54 rows , so we find the statistical data of Agent Rating
df["Agent_Rating"].describe()
```

```python
df["Agent_Rating"].median()
```

```python
print("non zero agent ratings are:")
# filling the missing values of Agent ratings 
df["Agent_Rating"] = df["Agent_Rating"].fillna(
    df["Agent_Rating"].median()
)
df["Agent_Rating"].value_counts()

```

```python
# Statistical information 
print(f"Statistical information of dataset: \n {df.describe()}")
```

```python
# checking for duplicates rows
print("Total duplicated rows: ",df.duplicated().sum())
# Checking about duplicated order ID
print("Total duplicated Order ID: ",df["Order_ID"].duplicated().sum())
```

```python
# check wheater the every id is unique or not 
print("Unique Order ID:", df["Order_ID"].nunique())
print("Total rows:", len(df))
```
